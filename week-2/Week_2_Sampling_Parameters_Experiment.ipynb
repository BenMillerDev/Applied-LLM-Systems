```python
!pip install google-genai -q

import os
from google.colab import userdata
from google import genai
from google.genai import types

os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
gemini_client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])
```

```python
MODEL = "gemini-3.1-flash-lite"
MAX_OUTPUT_TOKENS = 250
PROMPT = """Give one piece of advice for writing an effective prompt to an LLM."""

# Configuration 1: Temperature 0 -- greedy, near-deterministic decoding
response_temp_0 = gemini_client.models.generate_content(
    model=MODEL,
    contents=PROMPT,
    config=types.GenerateContentConfig(
        temperature=0,
        max_output_tokens=MAX_OUTPUT_TOKENS,
    ),
)
output_temp_0 = response_temp_0.text

# Configuration 2: Temperature 0.7, top_p 0.9 -- moderate randomness, nucleus-restricted
response_temp_07_top_p_09 = gemini_client.models.generate_content(
    model=MODEL,
    contents=PROMPT,
    config=types.GenerateContentConfig(
        temperature=0.7,
        top_p=0.9,
        max_output_tokens=MAX_OUTPUT_TOKENS,
    ),
)
output_temp_07_top_p_09 = response_temp_07_top_p_09.text

# Configuration 3: Temperature 1.2, top_k 50 -- high randomness, top 50 options considered
response_temp_12_top_k_50 = gemini_client.models.generate_content(
    model=MODEL,
    contents=PROMPT,
    config=types.GenerateContentConfig(
        temperature=1.2,
        top_k=50,
        max_output_tokens=MAX_OUTPUT_TOKENS,
    ),
)
output_temp_12_top_k_50 = response_temp_12_top_k_50.text

print("--- Temp 0 ---")
print(output_temp_0)
print()

print("--- Temp 0.7, top_p 0.9 ---")
print(output_temp_07_top_p_09)
print()

print("--- Temp 1.2, top_k 50 ---")
print(output_temp_12_top_k_50)
```
